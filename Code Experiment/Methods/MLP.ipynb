```python
import os
import copy
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    roc_auc_score, f1_score, classification_report,
    confusion_matrix, ConfusionMatrixDisplay, roc_curve
)
from sklearn.utils.class_weight import compute_class_weight

import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.regularizers import l1
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

# Kiểm tra GPU
gpu_devices = tf.config.list_physical_devices('GPU')
if gpu_devices:
    print(f"✅ Đã tìm thấy {len(gpu_devices)} GPU:")
    for gpu in gpu_devices:
        print(f"   - {gpu}")
    try:
        for gpu in gpu_devices:
            tf.config.experimental.set_memory_growth(gpu, True)
        print("✅ Đã bật Memory Growth cho GPU")
    except RuntimeError as e:
        print(e)
else:
    print("❌ Không tìm thấy GPU.")
    print("   → Vào Runtime > Change runtime type > Hardware accelerator: GPU")

np.random.seed(42)
tf.random.set_seed(42)

# ─────────────────────────────────────────────────────────────────────────────
# BƯỚC 1: LOAD DỮ LIỆU
# ─────────────────────────────────────────────────────────────────────────────
TRAIN_PATH = 'train_val_dataset_1e-4.txt'
TEST_PATH  = 'test_dataset_1e-4.txt'

def load_data(train_path, test_path):
    train_df = pd.read_csv(train_path, sep=' ')
    test_df  = pd.read_csv(test_path,  sep=' ')

    X_train_raw = train_df.drop(columns=['RID', 'Phenotype']).values
    y_train     = train_df['Phenotype'].values
    X_test_raw  = test_df.drop(columns=['RID', 'Phenotype']).values
    y_test      = test_df['Phenotype'].values

    return X_train_raw, y_train, X_test_raw, y_test

X_train_raw, y_train_full, X_test_raw, y_test = load_data(TRAIN_PATH, TEST_PATH)
N_var = X_train_raw.shape[1]

print(f"\n{'='*55}")
print(f"  Số mẫu train  : {X_train_raw.shape[0]}")
print(f"  Số mẫu test   : {X_test_raw.shape[0]}")
print(f"  Số SNP (p)    : {N_var}")
print(f"  Tỉ lệ p/n     : {N_var / X_train_raw.shape[0]:.1f}x  ← p >> n")
print(f"{'='*55}\n")

# ─────────────────────────────────────────────────────────────────────────────
# BƯỚC 2: CLASS WEIGHTS (xử lý mất cân bằng nhãn)
# ─────────────────────────────────────────────────────────────────────────────
classes = np.unique(y_train_full)
weights = compute_class_weight('balanced', classes=classes, y=y_train_full)
class_weight_dict = dict(zip(classes, weights))
print(f"Class weights: {class_weight_dict}")

# ─────────────────────────────────────────────────────────────────────────────
# BƯỚC 3: ĐỊNH NGHĨA KIẾN TRÚC MLP
#         (giữ nguyên so với bản gốc theo yêu cầu [FIX-3 = không thay đổi])
# ─────────────────────────────────────────────────────────────────────────────
def create_model(n_var, learning_rate=1e-4):
    inputs = Input(shape=(n_var,))

    x = Dropout(0.3)(inputs)
    x = Dense(2088, activation='sigmoid',
              kernel_regularizer=l1(5e-5))(x)
    x = Dropout(0.3)(x)

    x = Dense(696, activation='sigmoid',
              kernel_regularizer=l1(5e-5))(x)

    x = Dense(22, activation='sigmoid')(x)
    x = Dense(5,  activation='sigmoid')(x)
    outputs = Dense(1, activation='sigmoid')(x)

    model = Model(inputs, outputs)
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate),
        loss='binary_crossentropy',
        # [FIX-1] Thêm AUC vào metrics để EarlyStopping có thể theo dõi 'val_auc'
        metrics=[
            'accuracy',
            tf.keras.metrics.AUC(name='auc')
        ]
    )
    return model

# ─────────────────────────────────────────────────────────────────────────────
# BƯỚC 4: STRATIFIED 5-FOLD CROSS-VALIDATION
#         Mục tiêu: (a) đánh giá tổng quát hóa, (b) tìm best fold weights
# ─────────────────────────────────────────────────────────────────────────────
N_SPLITS   = 5
CV_EPOCHS  = 1000   # EarlyStopping sẽ dừng sớm — đây chỉ là giới hạn trên
BATCH_SIZE = 256

kfold      = StratifiedKFold(n_splits=N_SPLITS, shuffle=True, random_state=42)
cv_scores  = []       # val_auc của từng fold
best_epochs_list = [] # epoch thực tế dừng của từng fold
cv_histories = [] # NEW: To store training history for each fold

# Biến lưu thông tin best fold
best_fold_auc     = -1.0
best_fold_weights = None   # sẽ lưu get_weights() của model tốt nhất
best_fold_idx     = -1


for fold, (train_idx, val_idx) in enumerate(kfold.split(X_train_raw, y_train_full)):
    print(f"\n  ── Fold {fold + 1}/{N_SPLITS} ──")

    # Tách dữ liệu fold
    X_fold_train_raw = X_train_raw[train_idx]
    X_fold_val_raw   = X_train_raw[val_idx]
    y_fold_train     = y_train_full[train_idx]
    y_fold_val       = y_train_full[val_idx]

    scaler = StandardScaler()
    X_fold_train = scaler.fit_transform(X_fold_train_raw)
    X_fold_val   = scaler.transform(X_fold_val_raw)

    model_cv = create_model(N_var, learning_rate=1e-4)

    early_stop = EarlyStopping(
        monitor='val_auc',
        patience=30,
        restore_best_weights=True,
        mode='max',
        verbose=1
    )

    reduce_lr = ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=10,
        mode='min',
        min_lr=1e-6,
        verbose=1
    )

    history = model_cv.fit(
        X_fold_train, y_fold_train,
        validation_data=(X_fold_val, y_fold_val),
        epochs=CV_EPOCHS,
        batch_size=BATCH_SIZE,
        class_weight=class_weight_dict,
        callbacks=[early_stop, reduce_lr],
        verbose=0
    )

    cv_histories.append(history.history)


    if early_stop.stopped_epoch > 0:
        actual_best_epoch = early_stop.stopped_epoch - early_stop.patience + 1
    else:
        actual_best_epoch = CV_EPOCHS
    best_epochs_list.append(actual_best_epoch)

    val_preds_prob = model_cv.predict(X_fold_val, verbose=0)
    fold_auc = roc_auc_score(y_fold_val, val_preds_prob)
    cv_scores.append(fold_auc)

    print(f"     Epoch dừng thực tế : {actual_best_epoch}")
    print(f"     Val AUC            : {fold_auc:.4f}")


    if fold_auc > best_fold_auc:
        best_fold_auc     = fold_auc
        best_fold_weights = model_cv.get_weights()
        best_fold_idx     = fold + 1
        print(f"     ★ Best fold mới: Fold {best_fold_idx} (AUC={best_fold_auc:.4f})")

print(f"\n{'='*55}")
print(f"  KẾT QUẢ CROSS-VALIDATION")
print(f"  Average Val AUC : {np.mean(cv_scores):.4f} ± {np.std(cv_scores):.4f}")
print(f"  Best fold       : Fold {best_fold_idx} (AUC={best_fold_auc:.4f})")
print(f"  Avg best epoch  : {np.mean(best_epochs_list):.0f}")
print(f"{'='*55}\n")


FINETUNE_LR     = 1e-5
FINETUNE_EPOCHS = 50

print(f"{'='*55}")
print(f"  FINE-TUNE FINAL MODEL")
print(f"  Khởi tạo từ   : Fold {best_fold_idx} weights (AUC={best_fold_auc:.4f})")
print(f"  Dữ liệu       : Toàn bộ train set ({X_train_raw.shape[0]} mẫu)")
print(f"  LR fine-tune  : {FINETUNE_LR}")
print(f"  Epochs        : {FINETUNE_EPOCHS}")
print(f"{'='*55}\n")


final_scaler   = StandardScaler()
X_train_scaled = final_scaler.fit_transform(X_train_raw)
X_test_scaled  = final_scaler.transform(X_test_raw)


final_model = create_model(N_var, learning_rate=FINETUNE_LR)


final_model.set_weights(best_fold_weights)
print(f"✅ Đã nạp weights từ Fold {best_fold_idx}\n")
print(final_model.summary())

final_model.fit(
    X_train_scaled, y_train_full,
    epochs=FINETUNE_EPOCHS,
    batch_size=BATCH_SIZE,
    class_weight=class_weight_dict,
    verbose=1
)


print(f"\n{'='*55}")
print(f"  ĐÁNH GIÁ TRÊN TẬP TEST")
print(f"{'='*55}\n")

y_pred_probs = final_model.predict(X_test_scaled)
y_preds      = (y_pred_probs > 0.5).astype(int)

acc  = accuracy_score(y_test, y_preds)
prec = precision_score(y_test, y_preds, zero_division=0)
rec  = recall_score(y_test, y_preds, zero_division=0)
f1   = f1_score(y_test, y_preds, zero_division=0)
auc  = roc_auc_score(y_test, y_pred_probs)

print(f"  Accuracy  : {acc:.4f}")
print(f"  Precision : {prec:.4f}")
print(f"  Recall    : {rec:.4f}")
print(f"  F1-Score  : {f1:.4f}")
print(f"  AUC-ROC   : {auc:.4f}")
print(f"\nClassification Report:")
print(classification_report(y_test, y_preds,
      target_names=['Control (0)', 'Alzheimer (1)']))

# ─────────────────────────────────────────────────────────────────────────────
# BƯỚC 7: TRỰC QUAN HÓA
# ─────────────────────────────────────────────────────────────────────────────
os.makedirs('plots', exist_ok=True)

# ── 7.1 Confusion Matrix ─────────────────────────────────────────────────────
plt.figure(figsize=(6, 5))
ConfusionMatrixDisplay.from_predictions(
    y_test, y_preds,
    display_labels=['Control', 'Alzheimer'],
    cmap='Blues', values_format='d'
)
plt.title('Confusion Matrix — MLP v2')
plt.tight_layout()
plt.savefig('plots/mlp_v2_confusion_matrix.pdf', dpi=300, bbox_inches='tight')
plt.show()

# ── 7.2 ROC Curve ────────────────────────────────────────────────────────────
plt.figure(figsize=(7, 6))
fpr, tpr, _ = roc_curve(y_test, y_pred_probs)
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {auc:.4f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', label='Random')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve — MLP v2')
plt.legend(loc='lower right')
plt.tight_layout()
plt.savefig('plots/mlp_v2_roc_curve.pdf', dpi=300, bbox_inches='tight')
plt.show()

# ── 7.3 Risk Score Distribution ──────────────────────────────────────────────
plt.figure(figsize=(8, 6))
sns.kdeplot(y_pred_probs[y_test == 0].ravel(),
            label='Control', fill=True, color='green', alpha=0.5)
sns.kdeplot(y_pred_probs[y_test == 1].ravel(),
            label='Alzheimer', fill=True, color='red', alpha=0.5)
plt.xlabel('Predicted Probability')
plt.ylabel('Density')
plt.title('Risk Score Distribution — MLP v2')
plt.legend()
plt.tight_layout()
plt.savefig('plots/mlp_v2_risk_score_distribution.pdf', dpi=300, bbox_inches='tight')
plt.show()

# ── 7.4 CV AUC per Fold ──────────────────────────────────────────────────────
plt.figure(figsize=(7, 4))
fold_labels = [f'Fold {i+1}' for i in range(N_SPLITS)]
bar_colors  = ['gold' if i+1 == best_fold_idx else 'steelblue' for i in range(N_SPLITS)]
bars = plt.bar(fold_labels, cv_scores, color=bar_colors, edgecolor='black')
plt.axhline(np.mean(cv_scores), color='red', linestyle='--', label=f'Mean AUC = {np.mean(cv_scores):.4f}')
plt.ylim([0.0, 1.05])
plt.ylabel('Validation AUC')
plt.title('CV AUC per Fold — MLP v2\n(★ gold = best fold dùng làm khởi tạo)')
plt.legend()
plt.tight_layout()
plt.savefig('plots/mlp_v2_cv_auc_per_fold.pdf', dpi=300, bbox_inches='tight')
plt.show()

# ─────────────────────────────────────────────────────────────────────────────
# BƯỚC 8: LƯU KẾT QUẢ VÀ MÔ HÌNH
# ─────────────────────────────────────────────────────────────────────────────
os.makedirs('results', exist_ok=True)
os.makedirs('models', exist_ok=True)

# Lưu PRS scores
test_df_original = pd.read_csv(TEST_PATH, sep=' ')
prs_df = pd.DataFrame({
    'RID':           test_df_original['RID'].values,
    'Predicted_PRS': y_pred_probs.flatten()
})
prs_df.to_csv('results/patient_prs_scores_mlp_v2.csv', index=False)
print(f"\n✅ Đã lưu PRS scores → results/patient_prs_scores_mlp_v2.csv")
print(prs_df.head())

# Lưu trọng số model
final_model.save_weights('models/mlp_v2_finetuned.weights.h5')
print(f"✅ Đã lưu model weights → models/mlp_v2_finetuned.weights.h5")

# Lưu tóm tắt kết quả CV
cv_summary = pd.DataFrame({
    'Fold':       [f'Fold {i+1}' for i in range(N_SPLITS)],
    'Val_AUC':    cv_scores,
    'Best_Epoch': best_epochs_list,
    'Is_Best':    [i+1 == best_fold_idx for i in range(N_SPLITS)]
})
cv_summary.to_csv('results/cv_summary_mlp_v2.csv', index=False)
print(f"✅ Đã lưu CV summary → results/cv_summary_mlp_v2.csv")
print(cv_summary.to_string(index=False))
```

```python
print(f"\n{'='*55}")
print(f"  BẮT ĐẦU VẼ LEARNING CURVE")
print(f"{'='*55}")

# Khởi tạo lists để lưu trữ các giá trị của mỗi epoch trên tất cả các folds
all_fold_train_loss = []
all_fold_val_loss   = []
all_fold_train_auc  = []
all_fold_val_auc    = []

# Lặp qua từng fold history và trích xuất dữ liệu
for fold_history in cv_histories:
    all_fold_train_loss.append(fold_history['loss'])
    all_fold_val_loss.append(fold_history['val_loss'])
    all_fold_train_auc.append(fold_history['auc'])
    all_fold_val_auc.append(fold_history['val_auc'])

# Chuyển đổi sang numpy array để tính toán dễ hơn
# Đảm bảo các mảng có cùng chiều dài bằng cách chỉ lấy đến số epoch nhỏ nhất
min_epochs = min([len(h['loss']) for h in cv_histories])

all_fold_train_loss = np.array([h[:min_epochs] for h in all_fold_train_loss])
all_fold_val_loss   = np.array([h[:min_epochs] for h in all_fold_val_loss])
all_fold_train_auc  = np.array([h[:min_epochs] for h in all_fold_train_auc])
all_fold_val_auc    = np.array([h[:min_epochs] for h in all_fold_val_auc])

# Tính trung bình trên các folds
mean_train_loss = np.mean(all_fold_train_loss, axis=0)
mean_val_loss   = np.mean(all_fold_val_loss, axis=0)
mean_train_auc  = np.mean(all_fold_train_auc, axis=0)
mean_val_auc    = np.mean(all_fold_val_auc, axis=0)

epochs = range(1, min_epochs + 1)

# Vẽ biểu đồ Loss
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(epochs, mean_train_loss, 'b', label='Training Loss')
plt.plot(epochs, mean_val_loss, 'r', label='Validation Loss')
plt.title('Average Training and Validation Loss Across Folds')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

# Vẽ biểu đồ AUC
plt.subplot(1, 2, 2)
plt.plot(epochs, mean_train_auc, 'b', label='Training AUC')
plt.plot(epochs, mean_val_auc, 'r', label='Validation AUC')
plt.title('Average Training and Validation AUC Across Folds')
plt.xlabel('Epochs')
plt.ylabel('AUC')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.savefig('plots/mlp_v2_learning_curve.pdf', dpi=300, bbox_inches='tight')
plt.show()
print(f"\n✅ Đã vẽ Learning Curve và lưu → plots/mlp_v2_learning_curve.pdf")
```
